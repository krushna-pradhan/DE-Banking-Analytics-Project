### Creating Catalog and Schema

```python
%sql
-- CREATE BANKING CATALOG
CREATE CATALOG banking_catalog
Comment "Banking Analytics Catalog"
MANAGED LOCATION 'abfss://bronze@sabankinganalytics.dfs.core.windows.net/' ;

-- CREATE BANKING SCHEMA FOR MEDALION LAYERS
CREATE SCHEMA banking_catalog.bronze
COMMENT "Raw Banking Data"
MANAGED LOCATION 'abfss://bronze@sabankinganalytics.dfs.core.windows.net/' ;

CREATE SCHEMA banking_catalog.silver
COMMENT "Raw Banking Data"
MANAGED LOCATION 'abfss://silver@sabankinganalytics.dfs.core.windows.net/' ;

CREATE SCHEMA banking_catalog.gold
COMMENT "Raw Banking Data"
MANAGED LOCATION 'abfss://gold@sabankinganalytics.dfs.core.windows.net/' ;

--SHOW CATALOGS
SHOW CATALOGS;

--SHOW SCHEMAS
SHOW SCHEMAS IN banking_catalog;


```
