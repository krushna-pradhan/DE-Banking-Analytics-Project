### Read All Datasets from Raw Container


```python
%python
#Real all 10 banking CSV files from ADLS raw container

print("="*60)
print("Reading CSV files from ADLS")
print("="*60)

#Storage Account configuration 
storage_account = "sabankinganalytics"
raw_base_path = f"abfss://raw@{storage_account}.dfs.core.windows.net/"


#====================================================
#1. Reads Customers DataSet
#====================================================

print("\n1. Reading customers dataset")
customers_path = f"{raw_base_path}customers.csv/"
df_customers = spark.read.option("header", True).option("Inferschema",True).csv(customers_path)
print(f"customers records:  {df_customers.count()}")
print("Schema : ")
df_customers.printSchema()
display(df_customers.limit(5))

#====================================================
#2. Reads Accounts DataSet
#====================================================

print("\n1. Reading Accounts dataset")
accounts_path = f"{raw_base_path}accounts.csv/"
df_accounts = spark.read.option("header", True).option("Inferschema",True).csv(accounts_path)
print(f"Accounts records:  {df_accounts.count()}")
print("Schema : ")
df_accounts.printSchema()
display(df_accounts.limit(5))

#====================================================
#3. Reads Branches DataSet
#====================================================

print("\n1. Reading Branches dataset")
branches_path = f"{raw_base_path}branches.csv/"
df_branches = spark.read.option("header", True).option("Inferschema",True).csv(branches_path)
print(f"Branches records:  {df_branches.count()}")
print("Schema : ")
df_branches.printSchema()
display(df_branches.limit(5))

#====================================================
#4. Reads Credit cards DataSet
#====================================================

print("\n1. Reading Credit cards dataset")
credit_cards_path = f"{raw_base_path}credit_cards.csv/"
df_credit_cards = spark.read.option("header", True).option("Inferschema",True).csv(credit_cards_path)
print(f"Credit Card records:  {df_credit_cards.count()}")
print("Schema : ")
df_credit_cards.printSchema()
display(df_credit_cards.limit(5))

#====================================================
#5. Reads Customer Support Tickets DataSet
#====================================================

print("\n1. Reading Customer Support Tickets dataset")
customer_support_tickets_path = f"{raw_base_path}customer_support_tickets.csv/"
df_customer_support_tickets = spark.read.option("header", True).option("Inferschema",True).csv(customer_support_tickets_path)
print(f"Customer Support Tickets records:  {df_customer_support_tickets.count()}")
print("Schema : ")
df_customer_support_tickets.printSchema()
display(df_customer_support_tickets.limit(5))

#====================================================
#6. Reads Employees DataSet
#====================================================

print("\n1. Reading Employees dataset")
employees_path = f"{raw_base_path}employees.csv/"
df_employees = spark.read.option("header", True).option("Inferschema",True).csv(employees_path)
print(f"Empleyees records:  {df_employees.count()}")
print("Schema : ")
df_employees.printSchema()
display(df_employees.limit(5))

#====================================================
#7. Reads Fraud Transactions DataSet
#====================================================

print("\n1. Reading Fraud Transactions dataset")
fraud_transactions_path = f"{raw_base_path}fraud_transactions.csv/"
df_fraud_transactions = spark.read.option("header", True).option("Inferschema",True).csv(fraud_transactions_path)
print(f"Fraud Transactions records:  {df_fraud_transactions.count()}")
print("Schema : ")
df_fraud_transactions.printSchema()
display(df_fraud_transactions.limit(5))

#====================================================
#8. Reads Insurance Products DataSet
#====================================================

print("\n1. Reading Insurance Products dataset")
insurance_products_path = f"{raw_base_path}insurance_products.csv/"
df_insurance_products = spark.read.option("header", True).option("Inferschema",True).csv(insurance_products_path)
print(f"Insurance Products records:  {df_insurance_products.count()}")
print("Schema : ")
df_insurance_products.printSchema()
display(df_insurance_products.limit(5))

#====================================================
#9. Reads Loans DataSet
#====================================================

print("\n1. Reading Loans dataset")
loans_path = f"{raw_base_path}loans.csv/"
df_loans = spark.read.option("header", True).option("Inferschema",True).csv(loans_path)
print(f"Loans records:  {df_loans.count()}")
print("Schema : ")
df_loans.printSchema()
display(df_loans.limit(5))

#====================================================
#10. Reads Transactions DataSet
#====================================================

print("\n1. Reading Transactions dataset")
transactions_path = f"{raw_base_path}transactions.csv/"
df_transactions = spark.read.option("header", True).option("Inferschema",True).csv(transactions_path)
print(f"Transactions records:  {df_transactions.count()}")
print("Schema : ")
df_transactions.printSchema()
display(df_transactions.limit(5))

#====================================================
#Summary Reports :
#====================================================

print("\n" +"="*60)
print("DataSet Summary Report")
print("="*60)

summary_data = [
    ("customers", df_customers.count(), len(df_customers.columns)),
    ("accounts", df_accounts.count(), len(df_accounts.columns)),
    ("branches", df_branches.count(), len(df_branches.columns)),
    ("credit_cards", df_credit_cards.count(), len(df_credit_cards.columns)),
    ("customer_support_tickets", df_customer_support_tickets.count(), len(df_customer_support_tickets.columns)),
    ("employees", df_employees.count(), len(df_employees.columns)),
    ("fraud_transactions", df_fraud_transactions.count(), len(df_fraud_transactions.columns)),
    ("insurance_products", df_insurance_products.count(), len(df_insurance_products.columns)),
    ("loans", df_loans.count(), len(df_loans.columns)),
    ("transactions", df_transactions.count(), len(df_transactions.columns))
    
]

from pyspark.sql import Row
summary_df = spark.createDataFrame(
    [Row(Table=row[0], Record_count=row[1], Columns=row[2]) for row in summary_data]
)
display(summary_df)

print ("\n All 10 datasets loaded successfully")

```
